### Properties

- **Iterable**: elements can be accessed one at a time, for example using a `for` loop.
- **Indexed**: elements can be accessed by their position, such as `[0]`, `[1]`, etc.
- **Mutable**: the contents can be changed after the object has been created.
- **Callable**: the object can be called using parentheses `()`.

> **Note:** A dictionary is accessed by **key**, not by positional index.

### Categories

- **Scalar**: represents a single value.
- **Sequence**: contains an ordered sequence of elements.
- **Collection**: contains multiple elements, but is not necessarily ordered or position-indexed.
- **Mapped**: stores `key: value` pairs.
- **Container**: general term for an object that contains other objects.

> **Note:** A string is a sequence of characters. It is iterable and indexed, but immutable.

## Loops

### `for`

* A `for` loop is used to **iterate over the elements that is an iterable**.

* The loop runs through the iterable one element at a time and stops when there are no elements left.

* In a `for` loop, each element is temporarily assigned to a variable:

  ```python
  for element in my_list:
      do calculation
  ```

  The loop continues until the last element is reached.

* Remember that `str`, `list`, `dict`, `set`, and `tuple` are all **iterable**, so they can be used in a `for` loop.

For example:

```python

my_list = [1, 2, 3]

my_tuple = (1, 2, 3)

my_set = {1, 2, 3}

my_dict = {0: 1, 1: 2, 2: 3}

my_string = "name"

```

You can implement:

```python
for element in my_list:
    print(element)

for element in my_tuple:
    print(element)

for element in my_set:
    print(element)

for key in my_dict:
    print(key)

for character in my_string:
    print(character)
```




#### Iterating Over Dictionaries

A dictionary iterates over its **keys** by default.

For example:

```python
particles = {
    "particle_1": {
        "mass": 2.0,
        "velocity": 5.0
    },

    "particle_2": {
        "mass": 3.0,
        "velocity": 8.0
    }
}
```

Then:

```python
for element in particles:
    print(element)
```

will print the outer keys:

```python
particle_1
particle_2
```

To access both the keys and values of a dictionary, use `.items()`:

```python
print(particles.items())
```

Output:

```python
dict_items([
    ('particle_1', {'mass': 2.0, 'velocity': 5.0}),
    ('particle_2', {'mass': 3.0, 'velocity': 8.0})
])
```

Each item returned by `.items()` is a **key–value pair**:

```python
(key, value)
```

This is why two variables can be used when iterating over `.items()`:

```python
for key, value in particles.items():
    print(key, value)
```

Python assigns the key and value to the two variables `key` and `value`.

Output:

```python
particle_1 {'mass': 2.0, 'velocity': 5.0}
particle_2 {'mass': 3.0, 'velocity': 8.0}
```

Since `particles` is a nested dictionary, another `for` loop can be used to iterate through the inner dictionaries:

```python
for particle in particles:
    for key, value in particles[particle].items():
        print(particle, key, value)
```

Output:

```python
particle_1 mass 2.0
particle_1 velocity 5.0
particle_2 mass 3.0
particle_2 velocity 8.0
```



#### `range()`

A `for` loop can also iterate over a sequence of integers generated using `range()`:

```python
for index in range(start, stop, step):
    do calculation
```

where:

* `start` is the first value.
* `stop` is the value at which the range stops (**not included**).
* `step` is the increment between values.

For example:

```python
for index in range(0, 5, 1):
    print(index)
```

Output:

```python
0
1
2
3
4
```

If `start` and `step` are omitted, their default values are `0` and `1`:

```python
range(5)
```

is equivalent to:

```python
range(0, 5, 1)
```




#### `enumerate()`

`enumerate()` is used when both the **element** and its **counter/index** are needed.

The general syntax is:

```python
enumerate(iterable, start=0)
```

When used in a `for` loop:

```python
for index, element in enumerate(my_list):
    print(index, element)
```

For:

```python
my_list = [1, 2, 3]
```

the output is:

```python
0 1
1 2
2 3
```

The starting counter can be changed using `start`:

```python
for index, element in enumerate(my_list, start=1):
    print(index, element)
```

Output:

```python
1 1
2 2
3 3
```

`enumerate()` returns a counter and the corresponding element during each iteration.

**Note:** When `enumerate()` starts at `0`, the counter corresponds to the positional index for an indexed iterable. If `start` is changed, for example to `start=1`, the first value is a counter rather than the actual Python index.

#### Nests

To iterate over a nested dictionary, the general syntax is:

```python
for key1 in dictionary:
    for key2 in dictionary[key1]:
        value = dictionary[key1][key2]
        # work with value
```

For example,

```python
particles = {
    "particle_1": {
        "mass": 2.0,
        "velocity": 5.0
    },

    "particle_2": {
        "mass": 3.0,
        "velocity": 8.0
    }
}

for particle in particles:
    for key in particles[particle]:
        value = particles[particle][key]
        print(value)
```

will give:

```python
2.0
5.0
3.0
8.0
```

To print the value of a specific key:

```python
for particle in particles:
    value = particles[particle]["mass"]
    print(value)
```

To keep the same nested logic, you can use a combination of `for-if` statements:

```python
for particle in particles:
    for key in particles[particle]:
        value = particles[particle][key]

        if key == "mass":
            print(particles[particle][key])
```

You can apply the same logic to other nests by working from the outer to the inner nest. The general syntax is:

```python
for sublist_1 in my_list:
    for sublist_2 in sublist_1:
        for value in sublist_2:
            # operation with value
```

For example:

```python
matrix = [
    [1, 2, 3],  # outer_1: inner_1, inner_2, inner_3
    [4, 5, 6],  # outer_2: inner_1, inner_2, inner_3
    [7, 8, 9]   # outer_3: inner_1, inner_2, inner_3
]
```

To access the individual values, use:

```python
for outer in matrix:
    for inner in outer:
        print(inner)
```

The outer loop reads:

```python
outer = [1, 2, 3]
outer = [4, 5, 6]
outer = [7, 8, 9]
```

The inner loop then reads the individual values within each outer element:

```python
1
2
3
4
5

9
```

#### Data Storage

- In the above example, we saw that the loop creates single numbers and the data from each iteration are not stored in a collection.

- When the results from each iteration need to be stored, i.e., a complete list of the results, an empty collection can be created **before** the `for` loop.

- In the loop, the calculations are appended.

For example:

```python
my_list = [1, 2, 3]

result = []

for element in my_list:
    result.append(element * 2)

print(result)
```

Output:

```python id="4wymv2"
[2, 4, 6]
```

Here:

```python id="9k11lq"
result = []
```

is an empty list initialized before the loop so that each new result can be added to the same list, and the loop proceeds as:

```text id="s7wmcy"
result = []

element = 1  →  result = [2]

element = 2  →  result = [2, 4]

element = 3  →  result = [2, 4, 6]
```

NOTE the location of the `print` operation. If it is inside the loop, the current results will be printed after every iteration. 

Outside the loop, the complete results are printed after the loop has finished.

Also, if the empty list is created **inside** the loop, for example:

```python id="6s6k1y"
for element in my_list:
    result = []
    result.append(element * 2)
```

the list is reset during every iteration and only the result from the final iteration remains:

```python id="ng1fmo"
[6]
```

Thus, when accumulating results from a loop, the storage variable is generally **initialized before the loop**.

- Data can also be stored directly using a compressed loop expression called a **comprehension**:

```python id="r6gtj6"
output = [operation for variable in iterable if condition]

# or

output = [operation if condition else expression for variable in iterable]
```

Similar comprehension syntax can be used for different collections:

```python id="c6p8hd"
# List
output = [operation for variable in iterable]

# Set
output = {operation for variable in iterable}

# Dictionary
output = {key: operation for variable in iterable}

# Tuple
output = tuple(operation for variable in iterable)
```

- Statements such as:

```python id="uh8h36"
return
break
continue
```

cannot be used directly in the compressed form. Comprehensions are best used for simple operations.

Example:





````python
values = [0,1, 2, 3, 4]

square = [x**2 if x > 0 else None for x in values ]

print(square)
````

### `While` Loop

- remember that the condition is a `true` expression.

- a `while` loop can contain `break`, `contintue`, and `else`:

>> `break` terminate the loop.

>> `continue` skips the current iteration.

>> `else` tells the code what to do when the condition is not true. 


### Example

````python
counter = 6  # Start the counter at 0

while counter < 6:  # Keep looping while counter is less than 6

    counter += 1  # Increase counter by 1

    if counter == 2:
        continue  # Skip the rest of this iteration and return to while

    if counter == 5:
        break  # Exit the while loop completely

    print(counter)  # Print unless continue or break was executed

else:
    # Runs only if the while loop finishes normally (without break)
    print("Loop finished normally.")

# This is outside the while loop, so it always runs
print("Outside loop.")
````

Python allows you to create your own function using `def` in a temporary local environment. This organizes the code into reusable smaller sections and also prevents overwriting variables in the main code.

The variables and results inside a function exist only inside the function and are not created until the function is called. `return` sends the required value(s) from the function back to the main code. The returned value(s) can then be stored under new variables in the main code.

Assume

$$
a(n) = p\left(1 + \frac{r}{100}\right)^n
$$

where `r = 5` and `p = 100`.

```Python
def a(n):                       # This is called the header. `def` defines the function.
                                # `a` is the name of the function and `n` is the argument.
    p = 100
    r = 5                       # `p` and `r` are local variables.
                                # The indented statements make up the body of the function.
    return p * (1 + r/100)**n
```

In the function above, multiple arguments are separated by commas `(n1, n2, ...)`.

If there is no argument, the parentheses are left blank `()`.

Functions do nothing until called:

```Python id="0x8xfh"

Consider the double integral

$$
I = \int_0^1 \int_1^3 (x+y)\,dy\,dx
$$

The integral can be approximated numerically by dividing both $x$ and $y$ into $n$ intervals:

$$
\Delta x = \frac{1-0}{n},
\qquad
\Delta y = \frac{3-1}{n}
$$

and calculating

$$
I(n) \approx \sum_{i=0}^{n-1}\sum_{j=0}^{n-1}
(x_i+y_j)\Delta x\Delta y
$$

where

$$
x_i = i\Delta x,
\qquad
y_j = 1 + j\Delta y.
$$

1. Define a function `double_integral(n)` that calculates and returns $I$.

2. Use nested `for` loops to iterate over $i$ and $j$.

3. Calculate $x_i$ and $y_j$ inside the loops.

4. Add each contribution $(x_i+y_j)\Delta x\Delta y$ to the total.

5. Calculate and print the result for `n = 100`.

**Hint:** A double summation requires one loop inside another loop and initialization to start the summation.


````python
def double_integral(n):
    I = 0

    delta_x = (1 - 0) / n
    delta_y = (3 - 1) / n

    for j in range(n):
        for i in range(n):
            I += (i*delta_x + (1 + j*delta_y))*delta_x*delta_y

    return I

print(double_integral(100))

    
````


### Extra Challenge 2

The correlation between two variables $x$ and $y$ can be calculated using

$$
C_{xy} =
\frac{1}{N}
\sum_{i=1}^{N}
(x_i-\bar{x})(y_i-\bar{y})
$$

where $\bar{x}$ and $\bar{y}$ are the mean values of $x$ and $y$, and $N$ is the number of observations.

Consider the data:

```Python
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 7]
```

1. Define a function `correlation(x, y)` that takes the two lists as arguments.

2. Calculate the mean of each list:

```python
x_mean = sum(x) / len(x)
y_mean = sum(y) / len(y)
```

3. Initialize a variable to store the summation.

4. Use `zip(x, y)` in a `for` loop to access corresponding values $x_i$ and $y_i$.

5. For each pair, calculate

$$
(x_i-\bar{x})(y_i-\bar{y})
$$

and add it to the summation.

6. Divide the final summation by $N$.

7. Return the calculated value and print the result.

**Hint:** `zip(x, y)` groups corresponding values from the two lists:

```Python 
for x_i, y_i in zip(x, y):
```



````python
def correlation(x, y):
    x_mean = sum(x) / len(x)
    y_mean = sum(y) / len(y)

    Cxy = 0

    for x_i, y_i in zip(x,y):
        Cxy += (x_i - x_mean)*(y_i - y_mean)

    if len(x) == len(y):
        Cxy /= len(x)
    else:
        print('Size mismatch.')
        return

    return Cxy

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 7]

print(correlation(x,y))
````
