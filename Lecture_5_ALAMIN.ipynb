Functions take arguments as inputs, include documentation (a docstring) explaining what they do, and return outputs.

```python
def function_name(arg1, arg2, arg3):
    '''
    Documentation: here we explain what the function does. 
    This can be 
    on multiple
    lines
    '''
    # Here we can write the code that performs the desired operations using arg1, arg2, and arg3.
    return a,b,c # Return the results of the operations. (Optional)
```

```python
def hello(name):
    """prints 'hello' followed by the arg name"""
    print("hello,", name)

hello("cat")
```

Arguments can be input by order or name (keyword argument).

```python
def hello(name1,name2):
    """prints 'hello' followed by two  names"""
    print("hello,", name1, "and", name2)

hello("cat", "dog")                 #Call the function with two positional arguments

hello(name2="cat", name1="dog")     #Call the function with two keyword arguments (order does not matter)
```

You can also define functions with default argument values. For example:

```python
def add_numbers(num1, num2=2):
    """Returns the sum of two numbers. Second number has a default value of 2."""
    return num1 + num2

print(add_numbers(1, 3))  # Output: 4
print(add_numbers(1))     # Output: 3 (uses default value for num2)
```

An argument can be mostly any Python object, including numbers, strings, lists, dictionaries, and even other functions.

```python
def mean(numbers):
    """Returns the mean of a list of numbers."""
    return sum(numbers) / len(numbers)

mean_value = mean([1, 2, 3, 4, 5])  # Output: 3.0
```

You can also return multiple values.

```python
def mean_and_errors(numbers):
    """Returns the mean and standard error of a list of numbers."""
    import numpy as np
    mean_value = np.mean(numbers)
    std_error = np.std(numbers) / np.sqrt(len(numbers))
    return mean_value, std_error

numbers = [1, 2, 3, 4, 5]
mean_value, std_error = mean_and_errors(numbers)
print(f"Mean: {mean_value}, Standard Error: {std_error}")
```

Functions can lack a return statement, in which case they return None by default.

```python
import matplotlib.pyplot as plt
import numpy as np
def Add_logscales_xy():
    """This function add logscales to the x and y axes of a plot.
    It has no arguments and does not return any value. It modifies the current plot in place."""
    plt.xscale('log')  # Set x-axis to logarithmic scale
    plt.yscale('log')  # Set y-axis to logarithmic scale

plt.figure(figsize=(10, 3))  # Create a figure with a specified size
xs = np.logspace(0.1, 2, 100)  # Generate 100 points between 10^0.1 and 10^2
ys = xs**2  # y = x^2
plt.subplot(1, 2, 1)  # Create a subplot
plt.plot(xs, ys)  # Plot the data
plt.title('Linear Scale')  # Set title for the subplot
plt.subplot(1, 2, 2)  # Create another subplot
plt.plot(xs, ys)  # Plot the data
Add_logscales_xy()  # Call the function to add log scales to the axes
plt.title('Log Scale')  # Set title for the subplot
```

Activity 1: Write a function to return the two roots of a quadratic equation $a x^2 +b x +c =0$ only if the roots are real. If roots are complex numbers the function should print an error message and return None.

```python
def solve_quadratic(a,b,c):
    ''' This function takes three inputs a, b, and c, and returns the solution of ax^2 + bx + c = 0, provided that they are real.'''

    import numpy as np

    # checks if quadratic
    if a == 0:
        print('The equation is not quadratic.')
        return None, None
    
    # computes for discriminant
    disc = b**2 - 4*a*c
    # check the discriminant: b^2 - 4ac
    if disc < 0:
        print(f'Error: Discriminant is negative (D = {disc:.4f} < 0). Roots are complex.')
        return None, None

    xp = (-b + np.sqrt(disc)) / (2*a)
    xn = (-b - np.sqrt(disc)) / (2*a)

    # checks if discriminant is zero (one solution, mult 2)
    if disc == 0 and xp == xn:
        print(f'The equation is solved by x = {xp:.4f} (multiplicity 2).')
    else:
        print(f'The equation is solved by x = {xp:.4f} and x = {xn:.4f}.')

    return xp, xn

xp, xn = solve_quadratic(5, -2, -6)
```

Activity 2: Write a function that sets LaTeX style for the matplotlib plots. It should also set the figure dimensions to 10x3 inches by default, but such that the user can override this if needed. Also the user could set $x$ and $y$ log axis if desired (linear axis by default). The function should modify the current plot in place, just as in example . See Lecture_4.ipynb for the required matplotlib commands.

```python
import matplotlib.pyplot as plt
import numpy as np

def setup_plot(
    x = 10,
    y = 3
):
    '''This function sets LaTeX style for the matplotlib plots, and sets the figure dimensions (default: 10x3).'''

    # For LaTeX font
    plt.rcParams['mathtext.fontset'] = 'cm'
    plt.rcParams['font.family'] = 'serif'
    plt.rcParams['text.usetex'] = False
    plt.rcParams['font.family'] = 'serif'

    plt.figure(figsize=(x,y))

def add_logscales_xy(
        xlog = False,
        ylog = False
):
    '''This function adds logscales to x and y, if desired.'''

    if xlog:
        plt.xscale('log')
    if ylog:
        plt.yscale('log')


xs = np.logspace(0.1, 2, 100)  # Generate 100 points between 10^0.1 and 10^2
ys = xs**2  # y = x^2

x = 10
y = 3
setup_plot(x,y)  # new: setups the plot

plt.subplot(121)  # Create a subplot
plt.plot(xs, ys)  # Plot the data
plt.title('Linear Scale')  # Set title for the subplot

plt.subplot(122)  # Create another subplot
plt.plot(xs, ys)  # Plot the data

xlog = False
ylog = False
add_logscales_xy(xlog, ylog) # new: changes the logscale

# FOR PURPOSES OF THIS EXERCISE:
# Set title for the subplot as a conditional
if xlog and ylog:
    plt.title('Log-Log Scale')  
elif xlog and not ylog:
    plt.title('Lin-Log Scale')
elif not xlog and ylog:
    plt.title('Log-Lin Scale')
else:
    plt.title('Linear Scale')
    
```
