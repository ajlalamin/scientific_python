## Introduction

Python is a **dynamically typed, object-oriented language**.

An **object** is an entity that has:

- A **value** representing its data.
- **Attributes** defining characteristics associated with the object.
- **Methods** defining operations that can be performed on the object.

A **variable is a name that refers to an object**.

In Python, variables are created **dynamically and implicitly**.

> - **Implicit:** the type does not need to be explicitly declared as in C or Fortran.
> - **Dynamic:** a variable name can refer to objects of different types during execution.
> - **Duck typing:** the same operation can be applied to different types of objects if they support that operation.

### Python Objects

Many entities encountered in Python are objects.

| Object type | Definition | Example |
|---|---|---|
| **Data** | Represents a value or a collection of values | `9`, `3.14`, `"Hello"`, `[1, 2, 3]` |
| **Module** | Contains reusable Python code that can be imported | `math` — standard library; `numpy` — external package |
| **Function** | Performs an operation when called | `print`, `len` — built-in functions |
| **Class** | Defines a type of object and its associated attributes and methods | `class particle:` — user-defined class |
| **Method** | A function associated with an object or class | `my_list.append` — list method |
| **File object** | Provides access to an opened file | `open("data.txt")` — created using the built-in `open()` function |


### Built-in, Standard Library, and External

Python provides different sources:

| Source | Description | Example |
|---|---|---|
| **Built-in** | Available directly in Python without importing anything | `print()`, `len()`, `type()` |

| **External package** | Installed separately and then imported | `numpy`, `scipy`, `pandas` |
| **User-defined** | Created by the programmer | `particle`, `my_function()` |

### `print()` Function

The `print()` function displays the **string representation of an object or value** as output.

**Syntax:**

`print(object)`

| Example | Output |
|---|---|
| `print("Hello")` | `Hello` |
| `print(10)` | `10` |
| `print(x)` | Value of `x` |
| `print("x =", x)` | `x = 10` |
| `print(x, y)` | Values of `x` and `y` |
| `print(type(x))` | Type of object `x` |
| `print(x + y)` | Result of `x + y` |
| `print(x.upper())` | Result returned by the method |

> - `print()` can contain **values, variables, and expressions**.
> - Expressions inside `print()` are evaluated and their results are displayed, but the result is **not assigned to a variable**.
> - Statements such as assignment, `if`, `for`, and `del` cannot normally be used as arguments to `print()`.

### Basic Print Formatting

| Formatting | Purpose | Example |
|---|---|---|
| `,` | Print multiple objects separated by spaces | `print("x =", 10)` → `x = 10` |
| `sep=` | Change the separator between objects | `print(1, 2, 3, sep="-")` → `1-2-3` |
| `\n` | Start a new line | `print("Hello\nWorld")` |
| `\t` | Insert a tab | `print("x\ty")` |
| `f"..."` | Insert variables or expressions into a string | `print(f"x = {x}")` |

### f-String Formatting

An **f-string** inserts variables or expressions directly into a string using `{}`.

For example:

`print(f"x = {x}")`

Expressions can also be evaluated inside `{}`:

`print(f"x² = {x**2}")`

Numbers can be formatted using **format specifiers**:

| Format | Meaning | Example |
|---|---|---|
| `{x}` | Display the value | `f"{x}"` |
| `{x:.2f}` | Fixed-point with 2 decimal places | `3.14159` → `3.14` |
| `{x:.3f}` | Fixed-point with 3 decimal places | `3.14159` → `3.142` |
| `{x:.2e}` | Scientific notation with 2 decimal places | `1234` → `1.23e+03` |
| `{x:10.2f}` | Field width of 10 and 2 decimal places | Right-aligns the number in a field of width 10 |

## Variables

Variables are **symbols (names) that refer to objects**, which can be **scalars** or **collections**. The previous characteristics belong to the **object**, NOT the variable. A variable can be **reassigned** to another object. For example:

`x = 10`

then

`x = 20`

This does **not** change the object `10` into `20`. Instead, `x` is reassigned to a different object.
 
#### Variable Names

> - Variable names are **case-sensitive**.
> - May contain: `a-z`, `A-Z`, `0-9`, `_`.
> - Cannot start with a number.
> - Python does not impose a practical fixed length limit on variable names.

#### Naming Conventions

> - Variables and functions are normally written in **lowercase**.
> - Use **snake_case** for multiple words: `num_particles`.
> - Classes normally use **CapWords** (PascalCase): `Particle`, `ActiveParticle`.
> - A leading underscore `_name` indicates an **internal/non-public** name by convention.
> - Double-underscore names such as `__name__` are generally reserved for **special Python-defined names**.
> - Avoid using names of built-in functions: `print`, `list`, `sum`, `max`, etc.
> - Python **keywords** cannot be used as variable names: `for`, `if`, `class`, `return`, etc.


For example:

`x = 10`

`x = "ICTP"`

The name `x` first refers to an `int` object and is then reassigned to a `str` object.

### Operator Precedence

#### Which operation takes place first?

| Order | Operator | Operation | Direction |
|---|---|---|---|
| **1 — Highest** | `()` | Grouping / Parentheses | Innermost first |
| **2** | `**` | Exponentiation | **Right → Left** |
| **3** | `+x`, `-x` | Unary plus, unary minus | Right → Left |
| **4** | `*`, `/`, `//`, `%` | Multiplication, division, floor division, modulus | **Left → Right** |
| **5 — Lowest** | `+`, `-` | Addition, subtraction | **Left → Right** |



- Grouping `()` is evaluated first.

- Operators with the same precedence are generally evaluated **left to right**.

- Multiple exponentiation `**` e.g. 2^3^4 is an exception. It is evaluated from **right to left**.

- Nested parentheses are evaluated from the **innermost to the outermost**.



### Strings

Strings (`str`) are ordered sequences of characters.

> - Can be defined using single quotes `'...'` or double quotes `"..."` with the same result.  
>   Example: `'ICTP'` or `"ICTP"`
>
> - **Immutable** — cannot be changed after creation.
>
> - **Iterable** — characters can be accessed one at a time, for example using a `for` loop.  
>   Example: `"ICTP"` → `I`, `C`, `T`, `P`
>
> - **Ordered** — characters maintain their position and sequence.  
>   Example: `"ICTP"` → `I`, `C`, `T`, `P`
>
> - **Indexed** — characters can be accessed directly by their position (index).  
>   Indexing starts from `0`.  
>   Example: `"ICTP"[0]` → `"I"`

#### String Indexing

An **index (`idx`)** represents the position of a character in a string.

> - Positive indexing starts from `0` at the beginning.
> - Negative indexing starts from `-1` at the end.

For the string `"ICTP"`:

| Character | `I` | `C` | `T` | `P` |
|---|---:|---:|---:|---:|
| Positive index | `0` | `1` | `2` | `3` |
| Negative index | `-4` | `-3` | `-2` | `-1` |

Examples:

`"ICTP"[0]` → `"I"`  
`"ICTP"[1]` → `"C"`  
`"ICTP"[-1]` → `"P"`  
`"ICTP"[-2]` → `"T"`

#### String Slicing

`str[start_idx:end_idx:stride]`

| Parameter | Meaning | Example |
|---:|---:|---:|
| `start_idx` | Start at this index and continue to the end | `"ICTP"[1:]` → `"CTP"` |
| `end_idx` | Start at the beginning and stop before this index | `"ICTP"[:2]` → `"IC"` |
| `stride` | Select characters at a specified step | `"ICTP"[::2]` → `"IT"` |

- `start_idx` is included, while `end_idx` is excluded.

#### String Methods

Remember that strings are **immutable**, meaning that once a string object is created, its characters cannot be changed in place.

For example:

`name = "amna"`

followed by:

`name[0] = "A"`

will produce an error because the existing string cannot be modified.

String methods can be used to **create and return a new string with the desired changes**.

For example:

`name.capitalize()`

returns:

`"Amna"`

but the original value of `name` remains unchanged.

To keep the returned string, it can be reassigned:

`name = name.capitalize()`

Now:

`print(name)`

returns:

`Amna`

| `.split()` | Splits a string into a list | `x.split()` |
| `.strip()` | Removes leading and trailing whitespace | `x.strip()` |

> **Note:** String methods do not modify the original string. Methods that appear to change a string return a **new string object**.

### Booleans `bool`

### Lists `[]`

Lists are **collections of objects**. They are:

> - **Ordered:** items are stored in a defined order.
> - **Heterogeneous:** items do not have to be of the same data type.
> - **Indexed:** items are accessed by their index, starting from `0`.
> - **Mutable:** items can be added, removed, or changed after the list is created.


A list is created using square brackets `[]` and assigned to a variable, e.g.

`empty_list = []`

A list containing objects can be created as:

`my_list = [1, 2.0, "ICTP", True]`

`my_list[0]`

#### List of Lists

A **nested list** is a list that contains other lists. Nested lists can be used to represent **matrices, tables, or grouped data**.

Elements are accessed using **nested indexing**:

`nested_list[i][j]`

where:

- The first index `i` selects an **inner list or row**.
- The second index `j` selects an **element within that inner list or column**.

Similarly, a method requiring an argument can be applied to a selected inner list using:

`nested_list[i].method(value)`

For example, find the index of the value `5` in the second row.

**Answer:**

`nested_list[1].index(5)`

The first index `[1]` selects the second row, and `.index(5)` returns the index of the first occurrence of `5` within that row.


```python
a = ['a', 'b', 'c']       # Create a list of strings

n = [1, 2, 3, 4]          # Create a list of integers

x = [a, n, 3*[9]]             # Create a nested list containing a, n, and [9, 9, 9]

y = list(a) + list(n) + [9]*3  # Concatenate the elements of a and n with 9 into one list

print(x, len(x), sep="\n")                  # Output: [['a', 'b', 'c'], [1, 2, 3, 4], [9, 9, 9]]

print(y, len(y), sep="\n")                  # Output: ['a', 'b', 'c', 1, 2, 3, 4, 9, 9, 9] is one list
```

```python
print(x[1])  # The first index calls the whole sub-list

print(x[1][2:])  # The second index calls the elements of the sub-list
```

### Tuples `(,)`

Tuples are **ordered collections of objects**, similar to lists, but they are **immutable**.

> - **Ordered:** items are stored in a defined order.
> - **Indexed:** items are accessed by their index, starting from `0`.
> - **Heterogeneous:** items can have different data types.
> - **Immutable:** items cannot be added, removed, or changed after the tuple is created.
> - **Efficient:** tuples generally use less memory and can be slightly faster than lists.

A tuple is usually created using parentheses `()`:

`my_tuple = (1, 2.0, "ICTP", True)`

A single-item tuple requires a comma:

`single_tuple = (1,)`

Lists and tuples can be converted between each other using `list()` and `tuple()`:

`my_list = list(my_tuple)`

`my_tuple = tuple(my_list)`

This is useful when data need to be modified as a list and then stored as an immutable tuple.

In the example, `my_list = [1, 2.0, 'a', 'ciao']`, what can you do to prevent manipulation of the data?

```python
my_list = [1, 2.0, 'a', 'ciao']
my_tuple = tuple(my_list)
```

```python
my_tuple.append("Al Jay Lan Alamin")
```

To add an element, we can convert the tuple to a list so that it becomes mutable.

Choosing a list over a tuple for experimental data depends on the situation. If data is *currently* being collected and are added incrementally, then we must use a mutable object (easy to add / modify) like a list. However, if the data is already fixed and had already been collected by another researcher, then we should not easily be able to change the contents so we use a tuple.

```python
new_list = list(my_tuple)

new_list.append("Al Jay Lan Alamin")
print(new_list)
```

### Sets `{}`

> - **Unordered:** items do not have a defined position or index.

> - **Not indexed:** items cannot be accessed using an index such as `my_set[0]`.
> - **Immutable elements:** the objects stored in a set must be hashable (typically immutable), such as numbers, strings, and tuples.

`my_set = {1, 2, 3, 4}`

#### Common Set Operations

Let:

`B = {3, 4, 5, 6}`

### Dictionaries `{key: value}`

Dictionaries are **collections of key–value pairs**. They are:

> - **Key–value mappings:** each key is associated with a value.
> - **Insertion ordered:** items preserve the order in which they are added.
> - **Mutable:** key–value pairs can be added, removed, or changed.
> - **Unique keys:** each key can appear only once.
> - **Immutable keys:** keys must also be hashable, such as strings, numbers, boolean, or tuples. 
> - **Values:** values can be of any data type.

A dictionary can be created using curly brackets `{}` with `key: value` pairs:

`student = {"name": "Dorothy", "age": 25, "course": "Python", "grade": 85.5}`

Here:

| Key | Value |
|---|---|
| `"name"` | `"Dorothy"` |
| `"age"` | `25` |
| `"course"` | `"Python"` |
| `"grade"` | `85.5` |

A dictionary can also be created using `dict()`:

`student = dict(name="Dorothy", age=25, course="Python", grade=85.5)`

Values are accessed using their **keys**, rather than numerical indices:

`student["name"]`

returns:

`"Dorothy"`

`student["grade"]`

returns:

`85.5`

An empty dictionary is created using:

`empty_dict = {}`

or

`dict()`

#### Common Dictionary Methods

| Method | Description | Example |
|---|---|---|
| `.keys()` | Returns all keys | `student.keys()` |
| `.values()` | Returns all values | `student.values()` |
| `.items()` | Returns all key–value pairs | `student.items()` |
| `.get(key)` | Returns the value associated with a key | `student.get("name")` |
| `.update()` | Adds new or updates existing key–value pairs | `student.update({"grade": 90.0})` |
| `.pop(key)` | Removes a key–value pair and returns its value | `student.pop("age")` |
| `.clear()` | Removes all key–value pairs | `student.clear()` |
| `.copy()` | Creates a shallow copy of the dictionary | `new_student = student.copy()` |


### Example

```python
student = {"name": "Dorothy", "age": 25, "course": "Python", "grade": 85.5}

print(student.keys())                       # Returns keys of the dictionary

print(student.values())                     # Returns values of the dictionary    

print(student.items())                      # Retuns key-value pairs of the dictionary

student.update({"grade": 90.0})             # Updates the value of the key "grade" to 90.0
print(student)                              

student.update({"country": "Cameron"})      # Adds a new key-value pair to the dictionary
print(student)
```
